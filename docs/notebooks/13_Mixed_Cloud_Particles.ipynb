# Mixed Cloud Particles

This tutorial covers how to use mixed cloud particles within Virga and how to install the necessary dependencies. It is based on [Kiefer et al. in prep.]()

Clouds in substellar atmospheres can form from a variety of different materials. Some of these materials condense at very similar pressures, allowing the formation of mixed cloud particles. Our knowledge of the formation of mixed cloud particles is still minimal, but we do know that the mixing state affects observations [Kiefer et al. 2024](https://www.aanda.org/articles/aa/abs/2024/10/aa50526-24/aa50526-24.html). The Virga functionalities in discussed in this tutorial allow to account for the physics and opacities of well-mixed cloud particles.

## Step 1: Install MieAI

The opacity of mixed cloud particles depends on their mixing state. To handle this additional complexity, we developed MieAI [Attaway et al. in prep.](). This library allows to either fully calculate well-mixed Mie-theory, use grid interpolation, or AI models. The latter two significantly increase the computation time of mixed clouds within Virga. To install MieAI, follow these steps:

```
$ git clone https://github.com/d-attaway/mieai.git
$ cd mieai
$ pip install -e .
```

If possible it is recommended to use one of the pre-calcualted AI routines. To check if a model with your species is available, first download the available models, and then run the info function of MieAI:

````python
import mieai as MieAi

ma = MieAi()
ma.get_models()
ma.info()
````

If there is no suitable AI model, Virga defaults to full calculations, which can be rather slow. In this case, it is recommended to use grid interpolation. A detailed tutorial on how to set up grid interpolation can be found [in the MieAI tutorial](https://mieai.readthedocs.io/en/latest/index.html).

## Step 2: Run mixed clouds with Virga

Mixed cloud particles can be run exactly the same as a homogenous Virga cloud structure. All you have to do is set the `mixed' flag to true.

````python
import numpy as np
import pandas as pd
import astropy.units as u
import virga.justdoit as jdi
refrind_dir = '/Users/nbatalh1/Documents/data/virga'

# ==== Same example as in Running the Code but with 2 species
pressure = np.logspace(-5,3,30) #simple isotherml PT profile (kelvin)
temperature = np.zeros(30)+1600
kz = np.zeros(30)+ 1e10 #constant kz profile
metallicity = 1 #atmospheric metallicity relative to Solar
mean_molecular_weight = 2.2 # atmospheric mean molecular weight

# ==== Here you activate mixed cloud particles
sum_planet = jdi.Atmosphere(
    ['MgSiO3', 'SiO2'], fsed=0.1, mh=metallicity,
    mmw=mean_molecular_weight, mixed=True
)

# ==== Additional set up
sum_planet.gravity(gravity=357.00, gravity_unit=u.Unit('cm/(s**2)'))
sum_planet.ptk(df=pd.DataFrame({
    'pressure':pressure,
    'temperature':temperature,
    'kz':kz
}))

# ==== Run the code
````
